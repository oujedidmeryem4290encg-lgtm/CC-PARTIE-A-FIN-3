```python
# ===== Q2 — Cellule utilitaire (à exécuter UNE SEULE FOIS, avant les schémas) =====
import numpy as np, matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch, Ellipse, Circle
from sklearn.datasets import make_blobs, make_moons

rng = np.random.default_rng(42)
plt.rcParams.update({'figure.dpi': 110, 'axes.grid': True, 'grid.alpha': .25, 'font.size': 10})
C = {'bleu': '#2b6cb0', 'rouge': '#c53030', 'vert': '#2f855a', 'orange': '#dd6b20',
     'gris': '#4a5568', 'violet': '#6b46c1'}

def frontiere(ax, modele, X, y, titre='', s=28, cmap='coolwarm'):
    """Trace la frontière de décision d'un classifieur 2D + les points."""
    xx, yy = np.meshgrid(np.linspace(X[:, 0].min()-.7, X[:, 0].max()+.7, 250),
                         np.linspace(X[:, 1].min()-.7, X[:, 1].max()+.7, 250))
    zz = modele.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)
    ax.contourf(xx, yy, zz, alpha=.25, cmap=cmap)
    ax.scatter(X[:, 0], X[:, 1], c=y, cmap=cmap, edgecolor='k', s=s)
    ax.set_title(titre); ax.set_xlabel('x1'); ax.set_ylabel('x2')

def boite(ax, x, y, texte, w=2.4, h=1.1, couleur='#bee3f8', fs=9):
    ax.add_patch(FancyBboxPatch((x-w/2, y-h/2), w, h, boxstyle='round,pad=0.05',
                                fc=couleur, ec='#2d3748', lw=1.2))
    ax.text(x, y, texte, ha='center', va='center', fontsize=fs)

def fleche(ax, p1, p2, texte='', couleur='#2d3748', rad=0.0, fs=8):
    ax.annotate('', xy=p2, xytext=p1,
                arrowprops=dict(arrowstyle='->', lw=1.6, color=couleur,
                                connectionstyle=f'arc3,rad={rad}'))
    if texte:
        ax.text((p1[0]+p2[0])/2, (p1[1]+p2[1])/2+.25, texte, ha='center', fontsize=fs, color=couleur)

def flux(etapes, titre, formule=None, boucle=None, ax=None, couleurs=None):
    """Diagramme de flux horizontal. boucle=(i, j, 'texte') ajoute une flèche de retour de i vers j."""
    n = len(etapes)
    if ax is None:
        _, ax = plt.subplots(figsize=(2.9*n, 3.3))
    ax.axis('off'); ax.grid(False)
    ax.set_xlim(-1.5, 3*(n-1)+1.5); ax.set_ylim(-2.4, 1.6)
    for i, t in enumerate(etapes):
        boite(ax, 3*i, 0, t, couleur=(couleurs[i] if couleurs else '#bee3f8'))
        if i < n-1:
            fleche(ax, (3*i+1.3, 0), (3*(i+1)-1.3, 0))
    if boucle:
        i, j, txt = boucle
        ax.annotate('', xy=(3*j, -.65), xytext=(3*i, -.65),
                    arrowprops=dict(arrowstyle='->', lw=1.6, color=C['rouge'], connectionstyle='arc3,rad=-0.35'))
        ax.text(1.5*(i+j), -2.05, txt, ha='center', color=C['rouge'], fontsize=9)
    if formule:
        ax.text(1.5*(n-1), 1.25, formule, ha='center', fontsize=11)
    ax.set_title(titre, fontsize=11)
    return ax
```

## 1. Régression linéaire

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.linear_model import LinearRegression
modele=LinearRegression().fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Présentation

**Régression linéaire** (*Linear Regression, moindres carrés ordinaires*) — famille : régression supervisée. **Objectif** : prédire une variable numérique (ex. chiffre d'affaires) comme combinaison pondérée de variables explicatives.

**Intuition et étapes** : on construit la matrice des variables X, on cherche les coefficients qui minimisent la somme des carrés des erreurs sur l'entraînement, puis on applique la formule obtenue à de nouvelles observations.

**Formule / critère essentiel** :

$$\hat y_i=\beta_0+\sum_{j=1}^{p}\beta_j x_{ij}\qquad J(\beta)=\frac{1}{2n}\sum_{i=1}^{n}(y_i-\hat y_i)^2$$

**Symboles** : $n$ : nombre d'observations ; $p$ : nombre de variables ; $x_{ij}$ : valeur de la variable $j$ pour l'observation $i$ ; $y_i$ : valeur réelle ; $\hat y_i$ : prédiction ; $\beta_0$ : constante ; $\beta_j$ : coefficients ; $J$ : coût quadratique moyen.

**Paramètres appris** : $\beta_0,\beta_1,\dots,\beta_p$. **Hyperparamètres** : `fit_intercept` (inclure ou non la constante) ; pas de réglage de complexité.

**Avantages** : (1) rapide et facile à interpréter ; (2) sert de référence solide avant des modèles plus complexes.

**Limites** : (1) sensible aux valeurs atypiques et à la colinéarité ; (2) ne capte pas les relations non linéaires, et les coefficients ne prouvent aucune causalité.

### Schéma explicatif

**Comment lire ce schéma :** chaque point bleu est une observation (x_i, y_i). La droite rouge est le modèle ŷ = β0 + β1·x. Les petits segments gris verticaux sont les **résidus** (écart entre la vraie valeur et la prédiction). Les moindres carrés choisissent la droite qui rend la **somme des carrés de ces segments** minimale.
*Schéma original généré avec Matplotlib.*

```python
x = rng.uniform(0, 10, 40); y = 2*x + 1 + rng.normal(0, 2.5, 40)
b, a = np.polyfit(x, y, 1)
xs = np.linspace(0, 10, 50)
fig, ax = plt.subplots(figsize=(6.5, 4.2))
ax.scatter(x, y, c=C['bleu'], label='Observations $(x_i, y_i)$')
ax.plot(xs, a + b*xs, c=C['rouge'], lw=2, label=r'Droite ajustée $\hat{y}=\beta_0+\beta_1 x$')
ax.vlines(x, y, a + b*x, colors=C['gris'], lw=.9, label=r'Résidus $y_i-\hat{y}_i$ (à minimiser au carré)')
ax.set_xlabel('x (variable explicative)'); ax.set_ylabel('y (cible)')
ax.set_title('Régression linéaire : minimisation des résidus'); ax.legend(fontsize=8); plt.show()
```

1. Régression linéaire (Linear Regression)
Contextes d'utilisation:
1. Prévision du chiffre d'affaires d'une entreprise (Finance et gestion)

Une entreprise peut utiliser la régression linéaire pour prévoir son chiffre d'affaires mensuel en fonction de variables explicatives telles que les dépenses publicitaires, le nombre de clients, les prix des produits et les ventes précédentes.
- Données nécessaires : données historiques de ventes, dépenses publicitaires, prix et nombre de clients.
- Prérequis techniques : disposer de données numériques, d'une relation approximativement linéaire entre les variables explicatives et la variable cible, et d'un nombre suffisant d'observations.
2. Prévision des prix immobiliers (Commerce et immobilier)

Une agence immobilière peut utiliser cet algorithme pour estimer le prix d'un logement en fonction de sa superficie, du nombre de chambres, de son emplacement et de son ancienneté.
- Données nécessaires : prix historiques des logements, superficie, localisation, nombre de pièces et ancienneté.
- Prérequis techniques : disposer d'une variable cible numérique et de variables explicatives pertinentes. Un encodage des variables catégorielles est nécessaire.
Cas où l'algorithme serait peu approprié :

La régression linéaire serait peu adaptée à la prévision des ventes d'une entreprise lorsque celles-ci dépendent de relations fortement non linéaires, de changements brutaux du marché ou d'interactions complexes entre plusieurs variables. Dans ce cas, des modèles plus flexibles pourraient être envisagés.

Métriques d'évaluation :
- MAE (Mean Absolute Error) : mesure l'erreur absolue moyenne entre les valeurs réelles et prédites.
- RMSE (Root Mean Squared Error) : mesure l'erreur quadratique moyenne et pénalise davantage les grandes erreurs.
- R² (Coefficient de détermination) : mesure la proportion de la variance de la variable cible expliquée par le modèle.

## 2. Ridge

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),Ridge(alpha=1.0)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Présentation

**Régression Ridge** (*Ridge Regression, régularisation L2*) — famille : régression supervisée. **Objectif** : prédire une variable numérique en stabilisant la régression linéaire lorsque les variables sont corrélées.

**Intuition et étapes** : on standardise les variables, puis on minimise l'erreur quadratique plus une pénalité sur la taille des coefficients ; la force de la pénalité est choisie par validation croisée.

**Formule / critère essentiel** :

$$J(\beta)=\frac{1}{n}\lVert y-X\beta\rVert_2^2+\lambda\sum_{j=1}^{p}\beta_j^2$$

**Symboles** : $X$ : matrice $n\times p$ des variables ; $y$ : vecteur des cibles ; $\beta$ : coefficients (la constante n'est pas pénalisée) ; $\lambda$ : force de régularisation ; $\lVert\cdot\rVert_2$ : norme euclidienne.

**Paramètres appris** : les coefficients $\beta$. **Hyperparamètres** : `alpha` (= $\lambda$ ; plus il est grand, plus les coefficients sont rétrécis).

**Avantages** : (1) stable avec des variables fortement corrélées ; (2) réduit le surapprentissage.

**Limites** : (1) ne met pas les coefficients exactement à zéro (pas de sélection de variables) ; (2) exige la standardisation et le réglage de $\lambda$.

### Schéma explicatif

**Comment lire ce schéma :** chaque courbe est le coefficient d'une variable en fonction de la force de pénalisation α (échelle log, de gauche à droite : de plus en plus de régularisation). Avec Ridge, les coefficients **diminuent progressivement vers 0 mais ne sont jamais exactement nuls** : le modèle reste stable face aux variables corrélées.
*Schéma original généré avec Matplotlib.*

```python
from sklearn.datasets import make_regression
from sklearn.linear_model import Ridge
from sklearn.preprocessing import StandardScaler
X, y = make_regression(n_samples=200, n_features=6, n_informative=4, noise=10, random_state=42)
X = StandardScaler().fit_transform(X)
alphas = np.logspace(-2, 4, 60)
coefs = np.array([Ridge(alpha=al).fit(X, y).coef_ for al in alphas])
fig, ax = plt.subplots(figsize=(6.5, 4.2))
for j in range(coefs.shape[1]):
    ax.plot(alphas, coefs[:, j], label=f'coef. variable {j+1}')
ax.set_xscale('log'); ax.axhline(0, c='k', lw=.8)
ax.set_xlabel(r'Pénalisation $\alpha$ (échelle log)'); ax.set_ylabel('Valeur du coefficient')
ax.set_title(r'Ridge (pénalité L2) : les coefficients se rétractent vers 0'); ax.legend(fontsize=7, ncol=2); plt.show()
```

2. Ridge (Régression Ridge)
Contextes d'utilisation
1. Prévision des risques financiers d'une entreprise (Finance et gestion)

Une entreprise peut utiliser la régression Ridge pour prévoir ses dépenses ou ses résultats financiers à partir de plusieurs indicateurs tels que le chiffre d'affaires, les charges, les investissements et les dettes. Elle est particulièrement utile lorsque certaines variables explicatives sont fortement corrélées entre elles.
- Données nécessaires : données financières historiques, chiffre d'affaires, charges, investissements, dettes et autres indicateurs économiques.
- Prérequis techniques : disposer d'une variable cible numérique, de variables explicatives pertinentes et appliquer généralement une standardisation des variables numériques. Le choix du paramètre de régularisation alpha doit être effectué à l'aide d'une validation.
2. Prévision des prix immobiliers (Immobilier et commerce)

Une agence immobilière peut utiliser Ridge pour estimer les prix des logements en fonction de plusieurs caractéristiques : superficie, emplacement, nombre de chambres, ancienneté et proximité des services.
- Données nécessaires : prix historiques des logements et caractéristiques immobilières.
- Prérequis techniques : disposer de données structurées, d'une cible numérique et préparer les variables catégorielles par encodage. La standardisation est recommandée.
Cas où l'algorithme serait peu approprié :

La régression Ridge serait peu adaptée lorsque la relation entre les variables explicatives et la variable cible est fortement non linéaire et qu'aucune transformation appropriée n'est appliquée. Elle peut également être moins pertinente lorsque l'objectif principal est d'effectuer une sélection automatique des variables, car elle réduit leurs coefficients sans généralement les annuler complètement.

Métriques d'évaluation :
- MAE (Mean Absolute Error) : mesure l'erreur absolue moyenne entre les valeurs prédites et réelles.
- RMSE (Root Mean Squared Error) : permet de mesurer les erreurs en pénalisant davantage les grandes différences.
- R² (Coefficient de détermination) : mesure la proportion de variance de la variable cible expliquée par le modèle.

## 3. Lasso

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.linear_model import Lasso
modele=make_pipeline(StandardScaler(),Lasso(alpha=.1,max_iter=10000)).fit(Xr_ent,yr_ent)
print('R²:',r2_score(yr_test,modele.predict(Xr_test)))
```

### Présentation

**Régression Lasso** (*Lasso Regression, régularisation L1*) — famille : régression supervisée. **Objectif** : prédire une variable numérique tout en sélectionnant automatiquement les variables utiles (solution parcimonieuse).

**Intuition et étapes** : on standardise les variables, on minimise l'erreur quadratique plus la somme des valeurs absolues des coefficients ; certains coefficients deviennent exactement nuls ; on règle la pénalité par validation.

**Formule / critère essentiel** :

$$J(\beta)=\frac{1}{2n}\lVert y-X\beta\rVert_2^2+\lambda\sum_{j=1}^{p}\lvert\beta_j\rvert$$

**Symboles** : $n$ : taille de l'échantillon ; $X$ : matrice explicative ; $y$ : cible ; $\beta_j$ : coefficients ; $\lambda$ : force de la pénalité L1 ; $\lvert\cdot\rvert$ : valeur absolue.

**Paramètres appris** : les coefficients $\beta_j$ (dont beaucoup valent 0). **Hyperparamètres** : `alpha` (= $\lambda$), `max_iter` (itérations maximales de l'optimiseur).

**Avantages** : (1) sélection de variables intégrée, modèle plus lisible ; (2) efficace quand peu de variables sont réellement informatives.

**Limites** : (1) instable quand des variables sont très corrélées (il en retient une presque arbitrairement) ; (2) biaise les coefficients conservés vers zéro ; réglage de $\lambda$ nécessaire.

### Schéma explicatif

**Comment lire ce schéma :** même principe que pour Ridge, mais avec la pénalité L1. Quand α augmente, plusieurs courbes **tombent exactement à 0** et restent à 0 : le Lasso réalise une **sélection automatique de variables**. Les courbes qui survivent le plus longtemps correspondent aux variables les plus informatives.
*Schéma original généré avec Matplotlib.*

```python
from sklearn.datasets import make_regression
from sklearn.linear_model import Lasso
from sklearn.preprocessing import StandardScaler
X, y = make_regression(n_samples=200, n_features=6, n_informative=3, noise=10, random_state=42)
X = StandardScaler().fit_transform(X)
alphas = np.logspace(-2, 1.5, 60)
coefs = np.array([Lasso(alpha=al, max_iter=10000).fit(X, y).coef_ for al in alphas])
fig, ax = plt.subplots(figsize=(6.5, 4.2))
for j in range(coefs.shape[1]):
    ax.plot(alphas, coefs[:, j], label=f'coef. variable {j+1}')
ax.set_xscale('log'); ax.axhline(0, c='k', lw=.8)
ax.set_xlabel(r'Pénalisation $\alpha$ (échelle log)'); ax.set_ylabel('Valeur du coefficient')
ax.set_title(r'Lasso (pénalité L1) : coefficients annulés = variables éliminées'); ax.legend(fontsize=7, ncol=2); plt.show()
```

### Présentation

**Intuition et étapes** : on enrichit les variables avec leurs puissances et interactions (`PolynomialFeatures`), puis on ajuste un modèle linéaire (ici Ridge) sur cet espace élargi ; le degré est choisi par validation.

**Formule / critère essentiel** :

$$\hat y=\beta_0+\beta_1x+\beta_2x^2+\dots+\beta_d x^d$$

**Symboles** : $x$ : variable d'entrée ; $d$ : degré maximal ; $\beta_k$ : coefficients ; $\hat y$ : valeur estimée.

**Paramètres appris** : les coefficients $\beta_k$. **Hyperparamètres** : `degree` (degré $d$) et, dans l'exemple du notebook, `alpha` de la régularisation Ridge.

**Avantages** : (1) capte des courbures simples avec un modèle connu et rapide ; (2) reste interprétable pour un degré faible.

**Limites** : (1) le nombre de termes explose avec le degré et le nombre de variables ; (2) surapprend facilement et extrapole très mal hors du domaine observé.

Le degré est donc l'hyperparamètre clé.

## 5. Arbre de régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.tree import DecisionTreeRegressor
modele=DecisionTreeRegressor(max_depth=5,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Présentation

**Arbre de régression** (*Regression Tree*) — famille : régression supervisée. **Objectif** : prédire une valeur numérique en découpant l'espace des variables en zones homogènes (feuilles).

**Intuition et étapes** : on teste des variables et des seuils de coupure, on retient celui qui réduit le plus la dispersion de la cible, on répète récursivement dans chaque sous-ensemble ; la prédiction est la moyenne de la feuille atteinte.

**Formule / critère essentiel** :

$$\hat y(x)=\frac{1}{\lvert L(x)\rvert}\sum_{i\in L(x)} y_i$$

**Symboles** : $L(x)$ : observations d'entraînement dans la feuille contenant $x$ ; $\lvert L(x)\rvert$ : son effectif ; $y_i$ : cibles de ces observations.

**Paramètres appris** : variables et seuils de coupure, valeurs des feuilles. **Hyperparamètres** : `max_depth` (profondeur maximale), `min_samples_leaf` (taille minimale des feuilles).

**Avantages** : (1) capte les relations non linéaires et les interactions sans transformation ; (2) règles lisibles et visualisables.

**Limites** : (1) instable : un petit changement des données peut changer l'arbre ; (2) prédictions en escalier et extrapolation faible.

### Schéma explicatif

**Comment lire ce schéma :** l'arbre découpe l'axe x en intervalles (lignes verticales pointillées = **seuils de coupure**). Dans chaque intervalle, la prédiction est une **valeur constante** (la moyenne des y), d'où la courbe en escalier rouge. Plus la profondeur est grande, plus il y a de marches.
*Schéma original généré avec Matplotlib.*

```python
from sklearn.tree import DecisionTreeRegressor
x = np.sort(rng.uniform(0, 10, 80)); y = np.sin(x) + rng.normal(0, .25, 80)
arbre = DecisionTreeRegressor(max_depth=3, random_state=42).fit(x[:, None], y)
xs = np.linspace(0, 10, 500)
seuils = np.sort(arbre.tree_.threshold[arbre.tree_.feature >= 0])
fig, ax = plt.subplots(figsize=(6.5, 4.2))
ax.scatter(x, y, c=C['bleu'], s=22, label='Données')
ax.plot(xs, arbre.predict(xs[:, None]), c=C['rouge'], lw=2, label='Prédiction (constante par feuille)')
for i, t in enumerate(seuils): ax.axvline(t, c=C['gris'], ls=':', label='Seuils de coupure' if i == 0 else None)
ax.set_xlabel('x'); ax.set_ylabel('y'); ax.set_title('Arbre de régression (profondeur 3)'); ax.legend(fontsize=8); plt.show()
```

### Présentation

**Forêt aléatoire — régression** (*Random Forest Regressor*) — famille : régression supervisée (méthode d'ensemble). **Objectif** : prédire une valeur numérique en réduisant la variance d'un arbre unique par agrégation.

**Intuition et étapes** : on tire plusieurs échantillons bootstrap, on entraîne un arbre sur chacun avec un sous-ensemble aléatoire de variables à chaque coupure, puis on moyenne les prédictions des arbres.

**Formule / critère essentiel** :

$$\hat y(x)=\frac{1}{B}\sum_{b=1}^{B}T_b(x)$$

**Symboles** : $B$ : nombre d'arbres ; $T_b(x)$ : prédiction du $b$-ième arbre ; $x$ : nouvelle observation ; $\hat y$ : moyenne des prédictions.

**Paramètres appris** : la structure et les feuilles de chacun des $B$ arbres. **Hyperparamètres** : `n_estimators` ($B$), `max_depth`, `max_features`, `min_samples_leaf`.

**Avantages** : (1) robuste, bonne performance sans beaucoup de réglage ; (2) gère les non-linéarités et fournit une importance des variables.

**Limites** : (1) modèle volumineux, moins interprétable qu'un arbre seul ; (2) extrapole mal hors des valeurs vues à l'entraînement.

### Schéma explicatif

**Comment lire ce schéma :** les courbes fines colorées sont les prédictions de **15 arbres individuels**, entraînés sur des échantillons bootstrap différents : elles sont instables et très variables. La courbe noire épaisse est la **moyenne** de tous les arbres : elle est plus lisse, donc de **variance réduite**.
*Schéma original généré avec Matplotlib.*

```python
from sklearn.ensemble import RandomForestRegressor
x = np.sort(rng.uniform(0, 10, 100)); y = np.sin(x) + rng.normal(0, .3, 100)
xs = np.linspace(0, 10, 400)
foret = RandomForestRegressor(n_estimators=100, max_depth=5, random_state=42).fit(x[:, None], y)
fig, ax = plt.subplots(figsize=(6.5, 4.2))
ax.scatter(x, y, c='lightgray', edgecolor='gray', s=20, label='Données')
for i, t in enumerate(foret.estimators_[:15]):
    ax.plot(xs, t.predict(xs[:, None]), lw=.8, alpha=.6, label='Arbres individuels (15 sur 100)' if i == 0 else None)
ax.plot(xs, foret.predict(xs[:, None]), c='k', lw=2.5, label='Moyenne de la forêt')
ax.set_xlabel('x'); ax.set_ylabel('y'); ax.set_title('Forêt aléatoire : agrégation (bagging) d\'arbres'); ax.legend(fontsize=8); plt.show()
```

### Présentation

**Gradient Boosting — régression** (*Gradient Boosting Regressor*) — famille : régression supervisée (ensemble séquentiel). **Objectif** : obtenir une prédiction précise en additionnant de petits arbres qui corrigent progressivement les erreurs.

**Intuition et étapes** : on initialise avec une prédiction simple (souvent la moyenne), on calcule les résidus (pseudo-résidus de la perte), on entraîne un petit arbre pour les approcher, on l'ajoute avec un pas d'apprentissage, et on répète.

**Formule / critère essentiel** :

$$F_m(x)=F_{m-1}(x)+\eta\,h_m(x)$$

**Symboles** : $F_m$ : modèle à l'itération $m$ ; $h_m$ : nouvel arbre ajusté aux résidus ; $\eta$ : taux d'apprentissage ; $x$ : observation.

**Paramètres appris** : les arbres $h_m$ successifs. **Hyperparamètres** : `learning_rate` ($\eta$), `max_iter` (nombre d'arbres), `max_depth`, régularisation.

**Avantages** : (1) très performant sur données tabulaires ; (2) accepte plusieurs fonctions de perte.

**Limites** : (1) réglage sensible et risque de surapprentissage si trop d'itérations ; (2) entraînement séquentiel et peu interprétable.

### Schéma explicatif

**Comment lire ce schéma :** chaque panneau montre la prédiction du modèle après 1, 5 puis 100 arbres. Les arbres sont ajoutés **séquentiellement**, chacun corrigeant les erreurs (résidus) de l'ensemble précédent : la courbe rouge passe d'une grossière marche d'escalier à une approximation fine de la tendance.
*Schéma original généré avec Matplotlib.*

```python
from sklearn.ensemble import GradientBoostingRegressor
x = np.sort(rng.uniform(0, 10, 120)); y = np.sin(x) + rng.normal(0, .25, 120)
xs = np.linspace(0, 10, 400)
gb = GradientBoostingRegressor(n_estimators=100, max_depth=2, learning_rate=.3, random_state=42).fit(x[:, None], y)
etapes = list(gb.staged_predict(xs[:, None]))
fig, axs = plt.subplots(1, 3, figsize=(13, 3.8), sharey=True)
for ax, k in zip(axs, [1, 5, 100]):
    ax.scatter(x, y, c='lightgray', edgecolor='gray', s=18)
    ax.plot(xs, etapes[k-1], c=C['rouge'], lw=2); ax.set_title(f'Après {k} arbre(s)'); ax.set_xlabel('x')
axs[0].set_ylabel('y'); fig.suptitle('Gradient Boosting : correction progressive des résidus', y=1.02); plt.show()
```

## 8. Régression logistique

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.linear_model import LogisticRegression
modele=make_pipeline(StandardScaler(),LogisticRegression(max_iter=1000,random_state=GRAINE)).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Présentation

**Régression logistique** (*Logistic Regression*) — famille : classification supervisée. **Objectif** : estimer la probabilité d'appartenir à une classe (ex. défaut de paiement) et décider selon un seuil.

**Intuition et étapes** : on calcule un score linéaire des variables, on le transforme en probabilité par la sigmoïde, on ajuste les coefficients en minimisant la perte logarithmique, puis on classe selon un seuil (0,5 par défaut).

**Formule / critère essentiel** :

$$P(y=1\mid x)=\sigma(z)=\frac{1}{1+e^{-z}},\; z=\beta_0+\sum_j\beta_jx_j\qquad J=-\frac1n\sum_i\big[y_i\ln p_i+(1-y_i)\ln(1-p_i)\big]$$

**Symboles** : $x_j$ : variables ; $\beta_j$ : coefficients ; $z$ : score linéaire ; $\sigma$ : sigmoïde ; $p_i$ : probabilité prédite pour l'observation $i$ ; $y_i\in\{0,1\}$ : classe réelle ; $n$ : nombre d'observations.

**Paramètres appris** : $\beta_0$ et les $\beta_j$. **Hyperparamètres** : `C` (inverse de la régularisation), `penalty`, `max_iter`, seuil de décision.

**Avantages** : (1) sortie probabiliste et coefficients interprétables ; (2) rapide, bonne référence.

**Limites** : (1) frontière de décision linéaire seulement ; (2) sensible aux variables non standardisées/corrélées et au déséquilibre de classes.

### Schéma explicatif

**Comment lire ce schéma :** l'axe horizontal est la variable explicative, les points (0 ou 1) sont les classes observées. La courbe rouge en S est la **probabilité prédite P(y=1|x)** donnée par la fonction sigmoïde. La ligne pointillée à 0,5 est le **seuil de décision** : à droite du point où la courbe le croise, on prédit la classe 1.
*Schéma original généré avec Matplotlib.*

```python
from sklearn.linear_model import LogisticRegression
x = np.concatenate([rng.normal(3, 1.2, 40), rng.normal(6, 1.2, 40)]); y = np.r_[np.zeros(40), np.ones(40)]
m = LogisticRegression().fit(x[:, None], y)
xs = np.linspace(0, 9, 300); p = m.predict_proba(xs[:, None])[:, 1]
x_seuil = -m.intercept_[0] / m.coef_[0, 0]
fig, ax = plt.subplots(figsize=(6.5, 4.2))
ax.scatter(x, y, c=y, cmap='coolwarm', edgecolor='k', s=30, label='Observations (classe 0 ou 1)')
ax.plot(xs, p, c=C['rouge'], lw=2, label=r'$P(y=1|x)=\dfrac{1}{1+e^{-(\beta_0+\beta_1 x)}}$')
ax.axhline(.5, c=C['gris'], ls='--', label='Seuil 0,5'); ax.axvline(x_seuil, c=C['gris'], ls=':')
ax.set_xlabel('x'); ax.set_ylabel('Probabilité'); ax.set_title('Régression logistique : sigmoïde et seuil'); ax.legend(fontsize=8); plt.show()
```

8. Régression logistique (Logistic Regression)
Contextes d'utilisation
1. Prévision du défaut de paiement (Finance / Banque)
- Objectif : estimer la probabilité qu'un client rembourse ou non un crédit.
- Nature des données : revenus, historique de remboursement, montant du crédit, ancienneté bancaire et autres informations clients.
- Prérequis techniques : disposer d'une variable cible catégorielle, encoder les variables qualitatives, normaliser les variables numériques si nécessaire et traiter le déséquilibre des classes.

- Prérequis techniques : disposer d'exemples étiquetés et encoder les variables catégorielles.
Cas où l'algorithme est peu approprié :

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

**K plus proches voisins** (*K-Nearest Neighbors, KNN*) — famille : classification supervisée (à base d'instances). **Objectif** : classer une observation d'après la classe majoritaire de ses voisins les plus proches.

**Intuition et étapes** : on mémorise les données d'entraînement, on calcule la distance entre la nouvelle observation et chacune d'elles, on retient les $K$ plus proches, puis on prend le vote majoritaire.

**Formule / critère essentiel** :

$$\hat y(x)=\operatorname{mode}\{y_i : x_i\in V_K(x)\},\qquad d(x,x')=\sqrt{\sum_{j}(x_j-x'_j)^2}$$

**Symboles** : $V_K(x)$ : les $K$ plus proches voisins de $x$ ; $y_i$ : leurs classes ; $d$ : distance euclidienne ; $x_j$ : composante $j$.

**Paramètres appris** : aucun (le modèle est constitué des données mémorisées). **Hyperparamètres** : `n_neighbors` ($K$), métrique de distance, pondération des voisins.

**Avantages** : (1) très simple, sans phase d'apprentissage explicite ; (2) frontières de décision non linéaires.

**Limites** : (1) prédiction lente et gourmande en mémoire sur de gros volumes ; (2) sensible à l'échelle des variables (standardisation indispensable) et à la dimension élevée.

### Schéma explicatif

**Comment lire ce schéma :** le point étoilé noir est le nouvel individu à classer. Le cercle englobe ses **k = 5 plus proches voisins** (reliés par des traits). La classe prédite est le **vote majoritaire** parmi ces voisins. Changer k modifie la taille du cercle, donc la sensibilité au bruit.
*Schéma original généré avec Matplotlib.*

```python
from sklearn.neighbors import NearestNeighbors
X, y = make_blobs(n_samples=60, centers=2, cluster_std=1.7, random_state=3)
q = X.mean(axis=0, keepdims=True)
d, idx = NearestNeighbors(n_neighbors=5).fit(X).kneighbors(q)
votes = np.bincount(y[idx[0]], minlength=2)
fig, ax = plt.subplots(figsize=(5.5, 5.5))
ax.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm', edgecolor='k', s=35, label='Données d\'entraînement (2 classes)')
for i in idx[0]: ax.plot([q[0, 0], X[i, 0]], [q[0, 1], X[i, 1]], c=C['gris'], lw=1)
ax.add_patch(Circle(q[0], d.max(), fill=False, ls='--', ec=C['violet'], lw=1.5))
ax.scatter(*q[0], marker='*', s=300, c='gold', edgecolor='k', zorder=5, label='Nouveau point à classer')
ax.set_aspect('equal'); ax.set_xlabel('x1'); ax.set_ylabel('x2')
ax.set_title(f'KNN (k=5) : votes = classe 0 : {votes[0]}, classe 1 : {votes[1]}'); ax.legend(fontsize=8); plt.show()
```

## 10. Naive Bayes

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.naive_bayes import GaussianNB
modele=GaussianNB().fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Présentation

**Naïve Bayes gaussien** (*Gaussian Naive Bayes*) — famille : classification supervisée (probabiliste). **Objectif** : classer en appliquant le théorème de Bayes avec une hypothèse d'indépendance des variables sachant la classe.

**Intuition et étapes** : on estime la probabilité a priori de chaque classe et, pour chaque variable, sa moyenne et sa variance par classe ; pour une nouvelle observation, on multiplie les vraisemblances et on retient la classe la plus probable.

**Formule / critère essentiel** :

$$\hat y=\arg\max_c\; P(c)\prod_{j=1}^{p}\mathcal N(x_j\mid\mu_{jc},\sigma_{jc}^2)$$

**Symboles** : $c$ : classe ; $P(c)$ : probabilité a priori ; $x_j$ : variable $j$ ; $\mu_{jc},\sigma^2_{jc}$ : moyenne et variance de la variable $j$ dans la classe $c$ ; $p$ : nombre de variables.

**Paramètres appris** : $P(c)$, $\mu_{jc}$ et $\sigma^2_{jc}$. **Hyperparamètres** : `var_smoothing` (stabilisation numérique des variances).

**Avantages** : (1) entraînement et prédiction extrêmement rapides, peu de données nécessaires ; (2) probabilités par classe directement disponibles.

**Limites** : (1) l'indépendance des variables est rarement vraie ; (2) probabilités souvent mal calibrées, hypothèse gaussienne parfois inadaptée.

### Schéma explicatif

**Comment lire ce schéma :** chaque courbe est la densité gaussienne d'une variable pour une classe, **pondérée par la probabilité a priori** de la classe. Pour un x donné, Naive Bayes compare les deux hauteurs (règle de Bayes) et choisit la classe la plus haute. La droite verticale marque le **point de bascule** de la décision.
*Schéma original généré avec Matplotlib.*

```python
from scipy.stats import norm
mu0, s0, p0 = 3, 1.0, .6; mu1, s1, p1 = 6, 1.5, .4
xs = np.linspace(-1, 11, 500)
f0 = p0*norm.pdf(xs, mu0, s0); f1 = p1*norm.pdf(xs, mu1, s1)
x_bascule = xs[np.where(np.diff(np.sign(f0 - f1)))[0][-1]]
fig, ax = plt.subplots(figsize=(6.5, 4.2))
ax.plot(xs, f0, c=C['bleu'], lw=2, label=r'Classe 0 : $P(C_0)\,p(x|C_0)$')
ax.plot(xs, f1, c=C['rouge'], lw=2, label=r'Classe 1 : $P(C_1)\,p(x|C_1)$')
ax.fill_between(xs, f0, where=f0 >= f1, color=C['bleu'], alpha=.15); ax.fill_between(xs, f1, where=f1 > f0, color=C['rouge'], alpha=.15)
ax.axvline(x_bascule, c='k', ls='--', label=f'Frontière de décision (x ≈ {x_bascule:.1f})')
ax.set_xlabel('x'); ax.set_ylabel('Densité pondérée'); ax.set_title('Naive Bayes gaussien (1 variable) : règle de Bayes'); ax.legend(fontsize=8); plt.show()
```

## 11. Arbre de décision

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.tree import DecisionTreeClassifier
modele=DecisionTreeClassifier(max_depth=4,random_state=GRAINE).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Présentation

**Arbre de décision** (*Decision Tree Classifier*) — famille : classification supervisée. **Objectif** : classer par une suite de questions simples sur les variables, représentée sous forme d'arbre.

**Intuition et étapes** : à chaque nœud, on choisit la variable et le seuil qui réduisent le plus l'impureté des classes, on sépare les données, on répète jusqu'à un critère d'arrêt ; la classe d'une feuille est sa classe majoritaire.

**Formule / critère essentiel** :

$$G=1-\sum_{k=1}^{K}p_k^2\qquad \Delta G=G_{\text{parent}}-\Big(\tfrac{n_g}{n}G_g+\tfrac{n_d}{n}G_d\Big)$$

**Symboles** : $G$ : impureté de Gini d'un nœud ; $K$ : nombre de classes ; $p_k$ : proportion de la classe $k$ dans le nœud ; $n,n_g,n_d$ : effectifs du nœud et de ses fils gauche/droit ; $\Delta G$ : gain d'impureté.

**Paramètres appris** : variables et seuils de coupure, classes des feuilles. **Hyperparamètres** : `max_depth`, `min_samples_leaf`, `criterion` (gini ou entropie).

**Avantages** : (1) règles lisibles, sans standardisation nécessaire ; (2) gère les relations non linéaires.

**Limites** : (1) surapprend facilement s'il est trop profond ; (2) instable face à de petites variations des données.

### Schéma explicatif

**Comment lire ce schéma :** on lit l'arbre **de haut en bas**. Chaque nœud interne pose une question sur une variable (si vrai → branche gauche, sinon → branche droite). Chaque feuille donne la classe prédite ; la couleur indique la classe majoritaire, `gini` mesure l'impureté, `samples` le nombre d'exemples, `value` leur répartition par classe.
*Schéma original généré avec scikit-learn / Matplotlib (jeu Iris, pédagogique).*

```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier, plot_tree
iris = load_iris()
arbre = DecisionTreeClassifier(max_depth=2, random_state=42).fit(iris.data, iris.target)
fig, ax = plt.subplots(figsize=(9, 5))
plot_tree(arbre, feature_names=iris.feature_names, class_names=list(iris.target_names), filled=True, rounded=True, fontsize=9, ax=ax)
ax.set_title('Arbre de décision (profondeur 2) sur Iris'); plt.show()
```

### Présentation

**Forêt aléatoire — classification** (*Random Forest Classifier*) — famille : classification supervisée (méthode d'ensemble). **Objectif** : améliorer la fiabilité d'un arbre de décision en agrégeant de nombreux arbres différents.

**Paramètres appris** : la structure de chaque arbre. **Hyperparamètres** : `n_estimators` ($B$), `max_depth`, `max_features`, `class_weight`.

**Limites** : (1) moins interprétable et plus lourd qu'un arbre unique ; (2) probabilités parfois mal calibrées, classes rares défavorisées sans pondération.

### Schéma explicatif

Le dernier panneau montre la frontière de **la forêt** (vote majoritaire de 100 arbres), plus régulière et plus robuste au bruit.
*Schéma original généré avec Matplotlib.*

## 13. SVM

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.svm import SVC
modele=make_pipeline(StandardScaler(),SVC(kernel='rbf',C=1.0,random_state=GRAINE)).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Présentation

**Machine à vecteurs de support** (*Support Vector Machine, SVM*) — famille : classification supervisée. **Objectif** : trouver la frontière qui sépare les classes avec la plus grande marge.

**Intuition et étapes** : on standardise les variables, on cherche l'hyperplan à marge maximale en tolérant quelques erreurs ; un noyau (ex. RBF) permet des frontières non linéaires ; seuls les vecteurs de support définissent la frontière.

**Formule / critère essentiel** :

$$\min_{w,b,\xi}\ \tfrac12\lVert w\rVert^2+C\sum_i\xi_i\ \text{ s.c. } y_i\big(w^\top\phi(x_i)+b\big)\ge1-\xi_i,\quad K(x,x')=e^{-\gamma\lVert x-x'\rVert^2}$$

**Symboles** : $w,b$ : vecteur normal et biais de l'hyperplan ; $\xi_i\ge0$ : variables de relâchement (erreurs tolérées) ; $C$ : compromis marge/erreurs ; $y_i\in\{-1,1\}$ ; $\phi$ : transformation induite par le noyau $K$ ; $\gamma$ : largeur du noyau RBF.

**Paramètres appris** : $w$ et $b$ (ou coefficients des vecteurs de support). **Hyperparamètres** : `C`, `kernel`, `gamma`.

**Avantages** : (1) efficace en grande dimension et avec peu d'observations ; (2) frontières non linéaires grâce aux noyaux.

**Limites** : (1) coûteux sur de grands jeux de données et sensible à $C$ et $\gamma$ ; (2) pas de probabilités natives, difficile à interpréter.

### Schéma explicatif

**Comment lire ce schéma :** la ligne continue est l'**hyperplan séparateur** ; les deux lignes pointillées délimitent la **marge** (zone vide maximale). Les points entourés en noir sont les **vecteurs de support** : ils sont les seuls à déterminer la frontière, les autres points pourraient être supprimés sans la modifier.
*Schéma original généré avec Matplotlib (noyau linéaire pour la lisibilité).*

```python
from sklearn.svm import SVC
X, y = make_blobs(n_samples=60, centers=2, cluster_std=1.0, random_state=6)
svm = SVC(kernel='linear', C=1).fit(X, y)
xx, yy = np.meshgrid(np.linspace(X[:, 0].min()-1, X[:, 0].max()+1, 300), np.linspace(X[:, 1].min()-1, X[:, 1].max()+1, 300))
Z = svm.decision_function(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)
fig, ax = plt.subplots(figsize=(6.5, 5))
ax.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm', edgecolor='k', s=35)
ax.contour(xx, yy, Z, levels=[-1, 0, 1], colors='k', linestyles=['--', '-', '--'])
ax.scatter(*svm.support_vectors_.T, s=180, facecolors='none', edgecolors='k', lw=1.8, label='Vecteurs de support')
ax.plot([], [], 'k-', label='Hyperplan  w·x + b = 0'); ax.plot([], [], 'k--', label='Marge  w·x + b = ±1')
ax.set_xlabel('x1'); ax.set_ylabel('x2'); ax.set_title('SVM linéaire : marge maximale'); ax.legend(fontsize=8); plt.show()
```

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.ensemble import HistGradientBoostingClassifier
modele=HistGradientBoostingClassifier(max_iter=80,random_state=GRAINE).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Présentation

**Gradient Boosting — classification** (*Gradient Boosting Classifier*) — famille : classification supervisée (ensemble séquentiel). **Objectif** : obtenir un classifieur précis en additionnant des arbres qui corrigent les erreurs du modèle précédent.

**Intuition et étapes** : on part d'un score initial, on calcule le gradient négatif de la perte logarithmique, on entraîne un petit arbre dessus, on l'ajoute pondéré par un taux d'apprentissage, puis on convertit le score final en probabilité.

**Formule / critère essentiel** :

$$F_m(x)=F_{m-1}(x)+\eta\,h_m(x),\qquad p(x)=\sigma\big(F_M(x)\big)$$

**Symboles** : $F_m$ : score à l'itération $m$ ; $h_m$ : arbre ajusté au gradient négatif de la perte ; $\eta$ : taux d'apprentissage ; $\sigma$ : sigmoïde ; $M$ : nombre d'itérations ; $p$ : probabilité de la classe positive.

**Paramètres appris** : les arbres $h_m$. **Hyperparamètres** : `learning_rate`, `max_iter`, `max_depth`, régularisation L2.

**Avantages** : (1) très bonnes performances sur données tabulaires ; (2) capte interactions et non-linéarités automatiquement.

**Limites** : (1) nombreux hyperparamètres et risque de surapprentissage ; (2) moins interprétable et plus long à régler qu'un modèle linéaire.

### Schéma explicatif

**Comment lire ce schéma :** de gauche à droite, la frontière de décision du modèle après 1, 10 puis 100 arbres. Avec un seul arbre la frontière est grossière ; en ajoutant des arbres qui corrigent les erreurs de classement précédentes, elle épouse mieux la forme en « deux croissants » des données.
*Schéma original généré avec Matplotlib.*

```python
from sklearn.ensemble import GradientBoostingClassifier
X, y = make_moons(n_samples=300, noise=.25, random_state=42)
fig, axs = plt.subplots(1, 3, figsize=(13, 4))
for ax, n in zip(axs, [1, 10, 100]):
    m = GradientBoostingClassifier(n_estimators=n, max_depth=2, learning_rate=.3, random_state=42).fit(X, y)
    frontiere(ax, m, X, y, f'{n} arbre(s)')
fig.suptitle('Gradient Boosting : la frontière s\'affine à chaque itération', y=1.02); plt.show()
```

### Présentation

**AdaBoost** (*Adaptive Boosting*) — famille : classification supervisée (ensemble séquentiel). **Objectif** : combiner de nombreux classifieurs faibles en un classifieur fort en insistant sur les observations mal classées.

**Intuition et étapes** : on donne le même poids à toutes les observations, on entraîne un classifieur faible (souvent une souche), on augmente le poids des erreurs, on répète, puis on fait un vote pondéré par la qualité de chaque classifieur.

**Formule / critère essentiel** :

$$\alpha_m=\tfrac12\ln\frac{1-\varepsilon_m}{\varepsilon_m},\quad w_i\leftarrow w_i\,e^{-\alpha_m y_i h_m(x_i)},\quad H(x)=\operatorname{sign}\Big(\sum_m\alpha_m h_m(x)\Big)$$

**Symboles** : $h_m$ : classifieur faible $m$ ; $\varepsilon_m$ : son erreur pondérée ; $\alpha_m$ : son poids dans le vote ; $w_i$ : poids de l'observation $i$ ; $y_i\in\{-1,1\}$ ; $H$ : classifieur final.

**Paramètres appris** : les classifieurs faibles $h_m$ et leurs poids $\alpha_m$. **Hyperparamètres** : `n_estimators`, `learning_rate`, type de classifieur faible.

**Avantages** : (1) simple, souvent précis avec des apprenants très simples ; (2) peu d'hyperparamètres à régler.

**Limites** : (1) sensible au bruit et aux valeurs aberrantes (poids croissants sur les erreurs) ; (2) séquentiel, moins performant que le gradient boosting moderne.

### Schéma explicatif

**Comment lire ce schéma :** chaque panneau est une itération. La **taille des points** représente leur poids. Après chaque « souche » (arbre de profondeur 1), les points mal classés **grossissent** pour forcer le modèle suivant à s'y intéresser. Le titre indique l'erreur pondérée et le poids α accordé au classifieur faible.
*Schéma original généré avec Matplotlib (implémentation manuelle simplifiée d'AdaBoost).

```python
from sklearn.tree import DecisionTreeClassifier
X, y = make_moons(n_samples=80, noise=.25, random_state=1)
y_pm = 2*y - 1; w = np.ones(len(y)) / len(y)
fig, axs = plt.subplots(1, 3, figsize=(13, 4))
for t, ax in enumerate(axs):
    souche = DecisionTreeClassifier(max_depth=1, random_state=42).fit(X, y, sample_weight=w)
    pred = souche.predict(X); err = w[pred != y].sum(); alpha = .5*np.log((1-err)/err)
    frontiere(ax, souche, X, y, f'Itération {t+1} : erreur={err:.2f}, α={alpha:.2f}', s=w*len(y)*30)
    w = w*np.exp(-alpha*y_pm*(2*pred-1)); w /= w.sum()
fig.suptitle('AdaBoost : les points mal classés voient leur poids augmenter', y=1.02); plt.show()
```

### Présentation

**K-means (K-moyennes)** (*K-means*) — famille : clustering (non supervisé). **Objectif** : partitionner un ensemble d'observations en $K$ groupes (clusters) distincts de manière à ce que les données au sein d'un même groupe soient aussi similaires que possible (compacité).

**Intuition et étapes** :
1. Choix initial de $K$ centroïdes de manière aléatoire ou via K-means++.
2. Étape d'affectation : chaque point est associé au centroïde le plus proche (selon la distance euclidienne).
3. Étape de mise à jour : chaque centroïde est recalculé comme étant le barycentre (la moyenne) des points de son groupe.
4. Répétition des étapes 2 et 3 jusqu'à ce que les affectations ne changent plus ou que l'inertie converge.

**Formule / critère essentiel** :
$$J = \sum_{k=1}^{K} \sum_{x_i \in C_k} \lVert x_i - \mu_k \rVert^2$$

**Symboles** :
- $K$ : nombre de groupes défini à l'avance (hyperparamètre).
- $C_k$ : le $k$-ième cluster.
- $\mu_k$ : le centroïde du cluster $C_k$ (paramètre ajusté).
- $x_i$ : une observation.
- $\lVert \cdot \rVert$ : la distance euclidienne.
- $J$ : l'inertie intra-classe totale à minimiser.

**Avantages** :
1. Très rapide et performant sur des ensembles de données de grande taille.
2. Simple à comprendre et à implémenter.

**Limites** :
1. Nécessite de fixer le nombre de groupes $K$ à l'avance.
2. Très sensible à l'initialisation des centroïdes et aux valeurs aberrantes (outliers).

Chaque point a été rattaché au centre le plus proche.

## 18. Classification hiérarchique

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.cluster import AgglomerativeClustering
etiquettes=AgglomerativeClustering(n_clusters=3).fit_predict(X_clu)
print('Silhouette:',silhouette_score(X_clu,etiquettes))
```

### Présentation

**Classification hiérarchique ascendante (CAH)** (*Agglomerative Hierarchical Clustering*) — famille : clustering / partitionnement (non supervisé). **Objectif** : construire une hiérarchie arborescente de regroupements successifs (représentée sous forme de dendrogramme) à partir d'observations individuelles.

**Intuition et étapes** : au départ, chaque observation constitue un cluster unique. À chaque étape, l'algorithme calcule la matrice des distances ou de similarité entre toutes les paires de clusters existants et fusionne les deux clusters les plus proches selon un critère de liaison choisi. Ce processus itératif se poursuit jusqu'à ce que toutes les données soient agrégées en un seul cluster unique. Le nombre final de clusters est choisi a posteriori en effectuant une coupe horizontale de l'arbre (dendrogramme).

**Formule / critère de liaison (Exemple : critère de Ward)** :
$$\Delta(A, B) = \frac{|A| \cdot |B|}{|A| + |B|} \lVert \mu_A - \mu_B \rVert^2$$

**Symboles** :
- $A, B$ : deux clusters candidats à la fusion.
- $|A|, |B|$ : les effectifs respectifs (nombre d'observations) de ces deux clusters.
- $\mu_A, \mu_B$ : les vecteurs moyens (centroïdes) de chaque cluster.
- $\Delta(A, B)$ : l'augmentation d'inertie intra-classe provoquée par la fusion de $A$ et $B$ (que Ward cherche à minimiser à chaque étape).

**Paramètres appris** : l'arborescence des fusions (les liens d'affiliation).
**Hyperparamètres** : `n_clusters` (nombre final de groupes) ou `distance_threshold` (seuil de coupure), la métrique de distance (ex. euclidienne, Manhattan) et le critère de liaison (ex. `ward`, `complete`, `average`).

**Avantages** :
1. Ne requiert pas de spécifier le nombre de groupes $K$ au départ ; la décision du découpage peut être prise visuellement grâce au dendrogramme.
2. Permet d'explorer des relations de regroupement imbriquées et multiniveaux.

**Limites** :
1. Très coûteux en calculs et en mémoire ($O(n^2 \log n)$ au mieux, $O(n^3)$ au pire), ce qui le rend inadapté aux très grands jeux de données.
2. Les étapes de fusion sont irréversibles (si un point est mal regroupé au début, l'erreur se propage).

**Deux cas d'usage concrets :**

2. **Analyse de diversification sectorielle (Finance) :** Regrouper des portefeuilles ou des indices de marché par corrélations de performance afin d'identifier des structures hiérarchiques de secteurs d'investissement et optimiser l'allocation d'actifs.

- Score de Silhouette ou indice de Davies-Bouldin après choix de la coupure de l'arbre.

**Intuition et étapes** : chaque point forme d'abord un groupe ; à chaque étape on fusionne les deux groupes les plus proches selon un critère de liaison, jusqu'à n'en avoir plus qu'un ; on coupe l'arbre au niveau voulu.

**Symboles** : $A,B$ : deux groupes candidats à la fusion ; $\lvert A\rvert,\lvert B\rvert$ : leurs effectifs ; $\mu_A,\mu_B$ : leurs centres ; $\Delta$ : augmentation d'inertie provoquée par la fusion.

**Paramètres appris** : la suite des fusions (dendrogramme), pas de paramètres prédictifs. **Hyperparamètres** : `n_clusters` (ou `distance_threshold`), `linkage`, métrique de distance.

**Avantages** : (1) pas d'initialisation aléatoire, dendrogramme informatif ; (2) nombre de groupes choisi a posteriori.

**Limites** : (1) coût en temps et mémoire élevé (au moins quadratique en $n$) ; (2) fusions irréversibles, résultat dépendant du critère de liaison.

## 19. DBSCAN

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.cluster import DBSCAN
etiquettes=DBSCAN(eps=.8,min_samples=5).fit_predict(X_clu)
print('Groupes:',set(etiquettes),'Bruit:',(etiquettes==-1).sum())
```

### Présentation

**DBSCAN** (*Density-Based Spatial Clustering of Applications with Noise*) — famille : clustering (non supervisé, basé sur la densité). **Objectif** : trouver des groupes de forme quelconque dans les zones denses et isoler le bruit.

**Intuition et étapes** : pour chaque point on compte les voisins dans un rayon $\varepsilon$ ; un point dense est un point cœur ; les points cœurs voisins sont reliés en un même groupe, les points proches d'un cœur rejoignent le groupe, les autres sont du bruit (étiquette −1).

**Formule / critère essentiel** :

$$N_\varepsilon(x)=\{x' : d(x,x')\le\varepsilon\},\qquad x\text{ est un point cœur si }\lvert N_\varepsilon(x)\rvert\ge \text{minPts}$$

**Symboles** : $N_\varepsilon(x)$ : voisinage de $x$ ; $d$ : distance ; $\varepsilon$ : rayon du voisinage ; minPts : effectif minimal pour être dense.

**Paramètres appris** : les étiquettes de groupes (aucun centroïde). **Hyperparamètres** : `eps` ($\varepsilon$), `min_samples` (minPts), métrique de distance.

**Avantages** : (1) détecte des formes arbitraires sans fixer le nombre de groupes ; (2) identifie explicitement le bruit.

**Limites** : (1) difficile quand les densités des groupes varient fortement ; (2) très sensible à `eps`, peu efficace en dimension élevée.

### Présentation

**Mélanges gaussiens** (*Gaussian Mixture Model, GMM*) — famille : clustering (non supervisé, probabiliste). **Objectif** : modéliser les données comme un mélange de lois normales et affecter chaque point à un groupe avec une probabilité.

**Intuition et étapes** : on initialise les composantes, on alterne l'étape E (probabilité d'appartenance de chaque point à chaque composante) et l'étape M (mise à jour des poids, moyennes et covariances) jusqu'à convergence de la vraisemblance.

**Formule / critère essentiel** :

$$p(x)=\sum_{k=1}^{K}\pi_k\,\mathcal N(x\mid\mu_k,\Sigma_k),\qquad \sum_k\pi_k=1$$

**Symboles** : $K$ : nombre de composantes ; $\pi_k$ : poids de la composante $k$ ; $\mu_k$ : moyenne ; $\Sigma_k$ : matrice de covariance ; $\mathcal N$ : loi normale.

**Paramètres appris** : $\pi_k,\mu_k,\Sigma_k$. **Hyperparamètres** : `n_components` ($K$), `covariance_type`, initialisation.

**Avantages** : (1) affectations souples (probabilités) et groupes elliptiques de tailles différentes ; (2) modèle génératif, choix de $K$ possible par BIC/AIC.

**Limites** : (1) dépend de l'initialisation, optimum local possible ; (2) suppose des composantes gaussiennes, instable avec peu de données ou beaucoup de variables.

### Présentation

**Modèles de mélanges gaussiens** (*Gaussian Mixture Models - GMM*) — famille : clustering / partitionnement (non supervisé, approche probabiliste). **Objectif** : modéliser la distribution globale des données comme une combinaison de plusieurs distributions normales (gaussiennes) multidimensionnelles afin de réaliser un partitionnement souple (*soft clustering*).

**Intuition et étapes** :
L'algorithme suppose que chaque groupe suit une loi normale d'espérance (moyenne) $\mu_k$ et de matrice de covariance $\Sigma_k$. Pour ajuster ces paramètres, on utilise l'algorithme itératif **EM (Expectation-Maximization)** :
1. **Étape E (Expectation)** : calcule la probabilité (la responsabilité) que chaque point appartienne à chacune des composantes gaussiennes.
2. **Étape M (Maximization)** : met à jour les paramètres de chaque gaussienne (moyennes, covariances et poids du mélange) pour maximiser la vraisemblance des données.
Ces étapes sont répétées jusqu'à convergence de la log-vraisemblance.

**Formule / critère mathématique essentiel** :
$$p(x) = \sum_{k=1}^{K} \pi_k \mathcal{N}(x \mid \mu_k, \Sigma_k) \qquad \sum_{k=1}^{K} \pi_k = 1$$

**Symboles** :
- $K$ : nombre de composantes gaussiennes (hyperparamètre).
- $\pi_k$ : poids de la $k$-ième composante dans le mélange.
- $\mu_k$ : vecteur des moyennes de la composante $k$.
- $\Sigma_k$ : matrice de covariance (décrit la forme, l'orientation et l'étalement du groupe).
- $\mathcal{N}(x \mid \mu_k, \Sigma_k)$ : fonction de densité de la loi normale multidimensionnelle.

**Avantages** :
1. Permet un partitionnement "souple" (un point appartient à un groupe avec un score de probabilité entre 0 et 1).
2. Modélise des clusters de tailles et de géométries elliptiques variées grâce à la flexibilité de la matrice de covariance.

**Limites** :
1. Très sensible à l'initialisation des paramètres (peut converger vers des optimums locaux).
2. Risque de surapprentissage (singularity) si une composante ne contient que très peu de points de dimension élevée.

### Schéma explicatif

**Comment lire ce schéma :** Les ellipses de couleur représentent les contours de densité (à 1 et 2 écarts-types) pour chacune des trois composantes gaussiennes ajustées par le modèle GMM. Les points prennent une couleur hybride selon leurs probabilités d'appartenance respectives aux clusters (*soft clustering*), illustrant la transition douce entre les frontières de chaque groupe.
*Schéma original généré avec Matplotlib.*

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Ellipse
from sklearn.mixture import GaussianMixture

# Entraînement local de GMM
modele_gmm = GaussianMixture(n_components=3, random_state=GRAINE).fit(X_clu)
probabilites = modele_gmm.predict_proba(X_clu)

fig, ax = plt.subplots(figsize=(7, 4.5))

# Affichage des points avec couleur basée sur leurs probabilités d'appartenance
# On utilise une couleur mixte RGB représentant la contribution de chaque cluster
colors = probabilites
ax.scatter(X_clu[:, 0], X_clu[:, 1], c=colors, s=30, edgecolor='k', alpha=0.85, label='Points (soft clustering)')

# Fonction pour tracer l'ellipse de covariance
def tracer_ellipse_gmm(gmm, cluster_idx, ax, color, label):
    covariances = gmm.covariances_[cluster_idx]
    mean = gmm.means_[cluster_idx]
    v, w = np.linalg.eigh(covariances)
    u = w[0] / np.linalg.norm(w[0])
    angle = np.arctan2(u[1], u[0])
    angle = 180 * angle / np.pi
    v = 2.0 * np.sqrt(2.0) * np.sqrt(v)

    # Ellipse à 2 sigmas
    ellipse_2 = Ellipse(mean, v[0], v[1], angle=angle, edgecolor=color, fc='none', lw=2.0, ls='-', alpha=0.8, label=label)
    ax.add_patch(ellipse_2)
    # Ellipse à 1 sigma
    ellipse_1 = Ellipse(mean, v[0]*0.5, v[1]*0.5, angle=angle, edgecolor=color, fc='none', lw=1.0, ls='--', alpha=0.5)
    ax.add_patch(ellipse_1)
    # Centre
    ax.plot(mean[0], mean[1], 'o', color=color, markersize=8, markeredgecolor='black')

couleurs_groupes = ['red', 'green', 'blue']
for idx, col in enumerate(couleurs_groupes):
    tracer_ellipse_gmm(modele_gmm, idx, ax, col, f'Composante {idx+1} (̔_k)')

ax.set_title('Modèles de mélanges gaussiens (GMM) : Ellipses de covariance')
ax.set_xlabel('x1')
ax.set_ylabel('x2')
ax.legend(fontsize=8, loc='upper left')
plt.show()
```

### Contextes d'utilisation

**Deux cas d'usage concrets :**
1. **Segmentation comportementale floue (Marketing) :** Analyser le comportement des utilisateurs où un même client peut appartenir simultanément à plusieurs segments (ex. 60% client économe, 40% technophile) pour personnaliser les offres sans cloisonnement strict.
2. **Détection d'anomalies de performance industrielle (Logistique) :** Identifier des anomalies de fonctionnement sur des machines en calculant la probabilité que les mesures de capteurs proviennent du mélange des modes normaux d'opération. Si la densité de probabilité $p(x)$ est extrêmement faible, une alerte est déclenchée.

**Cas où l'algorithme est peu approprié :**
GMM est inadapté lorsque le volume de données est extrêmement faible par rapport au nombre de caractéristiques (dimension élevée), car le nombre de paramètres de la matrice de covariance à estimer explose en $O(p^2)$, menant à des instabilités de calcul importantes.

**Métriques adaptées :**
- Critère d'information bayésien (BIC) ou d'Akaike (AIC) pour sélectionner le nombre optimal de composantes $K$.
- Score de Silhouette probabiliste ou Log-vraisemblance moyenne.

## 21. ACP

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

**Analyse en composantes principales (ACP)** (*Principal Component Analysis - PCA*) — famille : réduction de dimension (non supervisé). **Objectif** : projeter un jeu de données à $p$ dimensions dans un espace de plus faible dimension $k < p$ tout en préservant le maximum d'information (variance) possible.

**Intuition et étapes** :
1. **Standardisation** : on centre et réduit les variables pour qu'elles aient le même poids.
2. **Matrice de covariance** : on calcule la corrélation et l'étalement conjoint de nos variables.
3. **Extraction** : par décomposition spectrale ou SVD (Singular Value Decomposition), on trouve les vecteurs propres (les axes principaux) et les valeurs propres (la variance capturée par chaque axe).
4. **Projection** : on projette les données initiales sur les axes principaux sélectionnés.

**Formule / critère mathématique essentiel** :
$$Z = X_c W_k \qquad \lambda_j w_j = \Sigma w_j$$

**Symboles** :
- $X_c$ : matrice des données centrées et réduites ($n \times p$).
- $W_k$ : matrice des $k$ premiers vecteurs propres de la matrice de covariance ($p \times k$).
- $Z$ : coordonnées projetées des points dans le nouvel espace ($n \times k$).
- $\Sigma$ : matrice de covariance des données de départ.
- $\lambda_j$ et $w_j$ : valeur propre et vecteur propre associés.

**Avantages** :
1. Élimine la multicolinéarité et réduit drastiquement le coût de calcul pour les modèles de machine learning en aval.
2. Permet la visualisation en 2D ou 3D de structures complexes initialement multidimensionnelles.

**Limites** :
1. Ne capture que les relations linéaires entre les variables explicatives.
2. Les nouvelles variables projetées (les composantes principales) sont des combinaisons linéaires des caractéristiques initiales et sont souvent très difficiles à interpréter concrètement.

### Schéma explicatif

**Comment lire ce schéma :** Le graphique montre le concept de l'ACP sur des données bidimensionnelles très corrélées. Le vecteur rouge représente le **premier axe principal (PC1)**, qui s'aligne selon la direction de variance maximale. Le vecteur vert est le **deuxième axe principal (PC2)**, orthogonal au premier. Les segments gris illustrent la projection orthogonale des points originaux sur PC1, montrant comment une dimension est préservée au mieux.
*Schéma original généré avec Matplotlib.*

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

# Génération de données synthétiques fortement corrélées
np.random.seed(42)
x_corr = np.random.normal(0, 1, 100)
y_corr = 1.5 * x_corr + np.random.normal(0, 0.4, 100)
X_corr = np.c_[x_corr, y_corr]

# Ajustement de l'ACP à 2 composantes
pca_2d = PCA(n_components=2).fit(X_corr)
mean = pca_2d.mean_
comp_1 = pca_2d.components_[0] * pca_2d.explained_variance_[0] * 1.5
comp_2 = pca_2d.components_[1] * pca_2d.explained_variance_[1] * 1.5

fig, ax = plt.subplots(figsize=(6.5, 5))
ax.scatter(X_corr[:, 0], X_corr[:, 1], c='#2b6cb0', alpha=0.7, edgecolors='k', label='Points de départ 2D')

# Tracé des composantes principales (axes de projection)
ax.annotate('', xy=mean + comp_1, xytext=mean, arrowprops=dict(arrowstyle='->', lw=2.5, color='#c53030', label='PC1 (Variance max)'))
ax.annotate('', xy=mean + comp_2, xytext=mean, arrowprops=dict(arrowstyle='->', lw=2.5, color='#2f855a', label='PC2 (Orthogonal)'))

# Visualisation de la projection sur PC1
pca_proj = PCA(n_components=1).fit(X_corr)
X_proj = pca_proj.inverse_transform(pca_proj.transform(X_corr))
for i in range(len(X_corr)):
    ax.plot([X_corr[i, 0], X_proj[i, 0]], [X_corr[i, 1], X_proj[i, 1]], color='gray', lw=0.6, ls='--')

ax.set_title('ACP : Recherche des directions de variance maximale')
ax.set_xlabel('Variable d\'origine x1')
ax.set_ylabel('Variable d\'origine x2')
ax.set_aspect('equal')
# Astuce pour la légende avec flèches
ax.plot([], [], color='#c53030', lw=2, label='PC1 (1ère Composante)')
ax.plot([], [], color='#2f855a', lw=2, label='PC2 (2ème Composante)')
ax.legend(fontsize=8)
plt.show()
```

### Contextes d'utilisation

**Deux cas d'usage concrets :**
1. **Compression d'images et reconnaissance faciale (Technologie / Logistique) :** Utiliser la technique des visages propres (*Eigenfaces*) pour réduire la dimensionnalité de grands volumes d'images de surveillance ou de reconnaissance d'identité avant de les classifier.
2. **Évaluation d'indicateurs macroéconomiques composites (Finance) :** Condenser des dizaines d'indices financiers corrélés (taux d'intérêt, inflation, taux de change, PIB) en un seul indicateur composite robuste (la première composante principale) résumant la santé du marché.

**Cas où l'algorithme est peu approprié :**
L'ACP est peu adaptée si les structures de données complexes sous-jacentes sont purement non linéaires (ex. distributions en forme de spirale ou de structures sphériques imbriquées). Dans ce type de scénario, des méthodes d'apprentissage de variétés (*manifold learning*) comme t-SNE, UMAP ou Kernel PCA donneront des résultats bien meilleurs.

**Métriques adaptées :**
- Ratio de variance expliquée cumulée.
- Valeurs propres individuelles (critère de Kaiser).

### Présentation

**Analyse en composantes principales (ACP)** (*Principal Component Analysis, PCA*) — famille : réduction de dimension (non supervisé). **Objectif** : résumer les données par quelques axes indépendants qui conservent le maximum de variance.

**Intuition et étapes** : on centre (et on standardise) les variables, on calcule la matrice de covariance, on en extrait les vecteurs propres classés par valeur propre décroissante, puis on projette les données sur les premiers axes.

**Formule / critère essentiel** :

$$Z=X_c\,W_k,\qquad \text{variance expliquée}_j=\frac{\lambda_j}{\sum_{l=1}^{p}\lambda_l}$$

**Symboles** : $X_c$ : données centrées ; $W_k$ : matrice des $k$ premiers vecteurs propres (composantes) ; $Z$ : coordonnées projetées ; $\lambda_j$ : valeur propre (variance portée par la composante $j$) ; $p$ : nombre de variables.

**Paramètres appris** : les moyennes, les composantes $W_k$ et les valeurs propres. **Hyperparamètres** : `n_components` (ou seuil de variance expliquée).

**Avantages** : (1) réduit la dimension, supprime la corrélation, facilite la visualisation ; (2) déterministe, rapide, réutilisable sur de nouvelles données.

**Limites** : (1) ne capte que des structures linéaires ; (2) sensible à l'échelle des variables, composantes parfois difficiles à interpréter.

### Présentation

**t-SNE** (*t-distributed Stochastic Neighbor Embedding* ou Plongement de voisins stochastiques distribué selon une loi t) — famille : réduction de dimension / apprentissage de variétés (non supervisé).

**Objectif et Intuition** : Projeter des données de grande dimension dans un espace de plus faible dimension (généralement 2D ou 3D) en préservant au maximum les relations de proximité locale (voisinages). Contrairement à l'ACP qui cherche à conserver la variance globale, le t-SNE s'assure que des points proches dans l'espace d'origine restent proches dans l'espace projeté.

**Étapes principales** :
1. Convertir les distances euclidiennes entre paires de points de l'espace d'origine en probabilités conditionnelles (gaussiennes) représentant leur similarité.
2. Définir une distribution de probabilité similaire (loi de Student ou t-distribution à un degré de liberté) pour les points projetés dans l'espace de basse dimension.
3. Minimiser la divergence de Kullback-Leibler (KL) entre ces deux distributions en utilisant une descente de gradient afin de déplacer les coordonnées projetées.

**Formule mathématique essentielle** :
$$\mathcal{L} = \mathrm{KL}(P \parallel Q) = \sum_{i \neq j} p_{ij} \ln \frac{p_{ij}}{q_{ij}}$$

**Symboles** :
- $p_{ij}$ : similarité jointe entre les points d'origine $x_i$ et $x_j$.
- $q_{ij} = \frac{(1 + \|y_i - y_j\|^2)^{-1}}{\sum_{k \neq l} (1 + \|y_k - y_l\|^2)^{-1}}$ : similarité jointe entre les projections basse dimension $y_i$ et $y_j$ (loi t de Student).
- $\mathcal{L}$ : fonction de coût (divergence de Kullback-Leibler) à minimiser par rapport aux coordonnées $y$.

**Paramètres appris** : Les coordonnées réelles $y_i$ de chaque point projeté (il n'y a pas de fonction mathématique globale de projection).
**Hyperparamètres clés** : `perplexity` (compromis entre l'attention portée aux voisinages locaux ou globaux), `learning_rate` et le nombre d'itérations d'optimisation.

**Avantages** :
1. Capacité exceptionnelle à démêler et visualiser des structures de données non linéaires hautement complexes.
2. Révèle des regroupements naturels (clusters) locaux invisibles avec des méthodes de projection linéaire.

**Limites** :
1. Pas de modèle de projection réutilisable : il est impossible de projeter directement un nouveau point sans relancer l'algorithme sur tout le jeu de données.
2. Ne conserve pas les distances globales : la distance entre clusters éloignés sur la carte 2D n'a pas de sens mathématique fiable.

### Schéma explicatif

**Comment lire ce schéma :** Nous générons un jeu de données en 3D en forme de rouleau de génoise (*Swiss Roll*). Nous appliquons d'une part l'ACP (projection linéaire) et d'autre part le t-SNE (projection non linéaire).

- Le **t-SNE** parvient à dérouler et séparer la variété en préservant le gradient de couleurs (les voisinages locaux de chaque couche).
*Schéma original généré avec Matplotlib.*

```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_swiss_roll
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.preprocessing import StandardScaler

# Génération d'un Swiss Roll 3D
X_swiss, couleur_swiss = make_swiss_roll(n_samples=800, noise=0.05, random_state=42)
X_swiss_std = StandardScaler().fit_transform(X_swiss)

# Projections
pca_proj = PCA(n_components=2).fit_transform(X_swiss_std)
tsne_proj = TSNE(n_components=2, perplexity=30, random_state=42).fit_transform(X_swiss_std)

# Tracé comparatif
fig = plt.figure(figsize=(15, 5))

# 1. Le Swiss Roll d'origine en 3D
ax1 = fig.add_subplot(1, 3, 1, projection='3d')
ax1.scatter(X_swiss[:, 0], X_swiss[:, 1], X_swiss[:, 2], c=couleur_swiss, cmap='Spectral', s=15)
ax1.set_title("Données d'origine 3D (Swiss Roll)")
ax1.view_init(10, -70)

# 2. Projection ACP (Linéaire)
ax2 = fig.add_subplot(1, 3, 2)
ax2.scatter(pca_proj[:, 0], pca_proj[:, 1], c=couleur_swiss, cmap='Spectral', s=15, edgecolor='k', alpha=0.7)
ax2.set_title("Projection linéaire : ACP")
ax2.set_xlabel("PC1")
ax2.set_ylabel("PC2")

# 3. Projection t-SNE (Non-linéaire)
ax3 = fig.add_subplot(1, 3, 3)
ax3.scatter(tsne_proj[:, 0], tsne_proj[:, 1], c=couleur_swiss, cmap='Spectral', s=15, edgecolor='k', alpha=0.7)
ax3.set_title("Projection non linéaire : t-SNE")
ax3.set_xlabel("Axe t-SNE 1")
ax3.set_ylabel("Axe t-SNE 2")

plt.tight_layout()
plt.show()
```

- **Conclusion** : Le t-SNE isole visuellement les différentes observations en formant des grappes distinctes et compactes dans l'espace projeté. Bien que les étiquettes de classes ne soient pas fournies à l'algorithme (non supervisé), la séparation nette des points confirme la pertinence du t-SNE pour capturer la structure intrinsèque et les sous-groupes de ce jeu de données.

### Présentation

**t-SNE** (*t-distributed Stochastic Neighbor Embedding*) — famille : réduction de dimension (non supervisé, visualisation). **Objectif** : projeter des données de grande dimension en 2D ou 3D en préservant les voisinages locaux.

**Intuition et étapes** : on calcule des similarités entre points dans l'espace d'origine (lois gaussiennes calibrées par la perplexité), on place les points en basse dimension avec des similarités de Student, puis on les déplace par descente de gradient pour réduire l'écart entre les deux.

**Formule / critère essentiel** :

$$\mathrm{KL}(P\Vert Q)=\sum_{i\ne j}p_{ij}\ln\frac{p_{ij}}{q_{ij}},\qquad q_{ij}\propto\big(1+\lVert y_i-y_j\rVert^2\big)^{-1}$$

**Symboles** : $p_{ij}$ : similarité de $x_i$ et $x_j$ dans l'espace d'origine ; $q_{ij}$ : similarité de leurs images $y_i,y_j$ ; $y_i$ : coordonnées réduites ; KL : divergence de Kullback-Leibler.

**Paramètres appris** : les coordonnées $y_i$ de chaque point (pas de modèle réutilisable). **Hyperparamètres** : `perplexity`, `learning_rate`, nombre d'itérations, initialisation.

**Avantages** : (1) révèle très bien des groupes locaux dans une carte 2D ; (2) capte des structures non linéaires.

**Limites** : (1) distances entre groupes et tailles des groupes non interprétables ; (2) sensible aux hyperparamètres, lent, pas de projection directe de nouvelles données.

## 23. UMAP

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
import umap
X_vin=StandardScaler().fit_transform(load_wine().data)
projection=umap.UMAP(n_components=2,n_neighbors=15,random_state=GRAINE).fit_transform(X_vin)
plt.scatter(projection[:,0],projection[:,1],s=10);plt.title('UMAP');plt.show()
```

### Présentation

**UMAP** (*Uniform Manifold Approximation and Projection*) — famille : réduction de dimension (non supervisé, visualisation). **Objectif** : projeter les données en basse dimension en conservant la structure locale et, mieux que t-SNE, une partie de la structure globale.

**Intuition et étapes** : on construit un graphe pondéré des $k$ plus proches voisins, on initialise une disposition en basse dimension, puis on l'optimise pour que son graphe ressemble à celui d'origine.

**Formule / critère essentiel** :

$$C=\sum_{i\ne j}\Big[\mu_{ij}\ln\frac{\mu_{ij}}{\nu_{ij}}+(1-\mu_{ij})\ln\frac{1-\mu_{ij}}{1-\nu_{ij}}\Big]$$

**Symboles** : $\mu_{ij}$ : poids de l'arête entre $i$ et $j$ dans l'espace d'origine ; $\nu_{ij}$ : poids correspondant dans l'espace réduit ; $C$ : entropie croisée à minimiser.

**Paramètres appris** : les coordonnées réduites (et le modèle permettant de projeter de nouveaux points). **Hyperparamètres** : `n_neighbors`, `min_dist`, `n_components`, métrique.

**Avantages** : (1) plus rapide que t-SNE, avec une méthode `transform` pour de nouvelles données ; (2) préserve souvent mieux la structure globale.

**Limites** : (1) résultats fortement dépendants de `n_neighbors` et `min_dist` ; (2) distances et tailles de groupes à interpréter avec prudence.

### Présentation

**Projection et Approximation de Variétés Uniformes** (*Uniform Manifold Approximation and Projection - UMAP*) — famille : réduction de dimension / apprentissage de variétés (non supervisé).

**Objectif et Intuition** : Projeter des données de haute dimension dans un espace de plus faible dimension (ex. 2D) en préservant à la fois la structure locale (voisinages proches) et la structure globale du jeu de données. L'intuition repose sur la géométrie riemannienne et la topologie algébrique : UMAP suppose que les données sont distribuées uniformément sur une variété locale et cherche à reproduire la structure de cette variété sous forme de graphe flou en basse dimension.

**Étapes principales** :
1. Construire une représentation sous forme de graphe flou des plus proches voisins dans l'espace de haute dimension (calcul de la force des connexions).
2. Initialiser la projection en basse dimension en utilisant une décomposition spectrale de graphe (pour préserver la structure globale).
3. Optimiser la position des points dans l'espace cible de faible dimension en minimisant l'entropie croisée floue entre les représentations de haute et basse dimension à l'aide d'une descente de gradient stochastique.

**Formule / critère mathématique essentiel** :
$$C_{\mathrm{UMAP}} = \sum_{i \neq j} \left[ \mu_{ij} \ln \frac{\mu_{ij}}{\nu_{ij}} + (1 - \mu_{ij}) \ln \frac{1 - \mu_{ij}}{1 - \nu_{ij}} \right]$$

**Symboles** :
- $\mu_{ij}$ : similarité (poids de l'arête) entre les points $x_i$ et $x_j$ dans l'espace d'origine.
- $\nu_{ij} = (1 + a \|y_i - y_j\|^{2b})^{-1}$ : similarité entre les projections $y_i$ et $y_j$ dans l'espace réduit (où $a$ et $b$ sont des coefficients ajustés par rapport à l'hyperparamètre `min_dist`).
- $C_{\mathrm{UMAP}}$ : entropie croisée floue (coût global à minimiser).

**Paramètres appris** : Les coordonnées réelles $y_i$ de chaque observation projetée. L'algorithme apprend également un modèle de projection capable de transformer de nouvelles données.
**Hyperparamètres** : `n_neighbors` (taille du voisinage local), `min_dist` (distance minimale entre les points projetés), `metric` (mesure de distance initiale).

**Avantages** :
1. Conserve nettement mieux la structure globale des données que le t-SNE tout en étant extrêmement rapide.
2. Permet de projeter de nouvelles observations inconnues sur une carte existante (méthode `.transform()`).

**Limites** :
1. Tracé dépendant des choix combinés de `n_neighbors` et `min_dist`.
2. Les échelles réelles des distances sur les axes projetés n'ont pas de signification physique directe.

```python
from sklearn.ensemble import IsolationForest
X_anomal=np.vstack([X_clu,np.array([[9,9],[-8,-8]])])
modele=IsolationForest(contamination=.02,random_state=GRAINE).fit(X_anomal)
print('Observations atypiques:',np.where(modele.predict(X_anomal)==-1)[0])
```

### Présentation

**Forêt d'isolement** (*Isolation Forest*) — famille : détection d'anomalies (non supervisé). **Objectif** : repérer et isoler les observations atypiques (anomalies) au sein d'un jeu de données sans utiliser d'étiquettes de classification.

**Intuition et étapes** : l'algorithme repose sur le principe que les anomalies sont rares et situées loin dans l'espace des caractéristiques. Il construit un ensemble d'arbres de décision aléatoires (arbres d'isolement) où, à chaque nœud, une variable et un seuil de coupure sont choisis de façon purement aléatoire. Les anomalies, étant isolées plus facilement que les points normaux, se retrouvent à des profondeurs de feuilles très courtes.

**Formule / critère essentiel** :
$$s(x, n) = 2^{-\frac{E(h(x))}{c(n)}}$$

**Symboles** :
- $s(x, n)$ : score d'anomalie du point $x$ pour un échantillon de taille $n$ (compris entre 0 et 1).
- $h(x)$ : longueur du chemin (profondeur) pour isoler $x$ dans un arbre d'isolement.
- $E(h(x))$ : espérance (moyenne) de la longueur du chemin sur l'ensemble de la forêt.
- $c(n) = 2\ln(n-1) + 0.5772156649 - \frac{2(n-1)}{n}$ : longueur moyenne d'un chemin de recherche infructueux dans un arbre binaire de recherche (utilisé pour normaliser $E(h(x))$).

**Paramètres appris** : La structure de partitionnement aléatoire de chaque arbre d'isolement.
**Hyperparamètres** : `n_estimators` (nombre d'arbres), `max_samples` (nombre de points échantillonnés par arbre) et `contamination` (proportion attendue d'anomalies).

**Avantages** :
1. Très performant sur de grands volumes de données grâce à une complexité linéaire en temps.
2. Ne nécessite aucune phase d'entraînement étiquetée (apprentissage non supervisé).

**Limites** :
1. Le choix de l'hyperparamètre `contamination` influence directement le taux de fausses alertes.
2. Moins efficace si les anomalies forment des clusters denses de tailles significatives.

Plus un point est isolé des régions à forte densité, plus la ligne de niveau de contour s'assombrit (indiquant un score d'anomalie élevé proche de 1).

## 25. Bandits manchots

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
generateur=np.random.default_rng(GRAINE)
vraies_probas=np.array([.2,.5,.7]);somme=np.zeros(3);comptes=np.zeros(3)
for t in range(500):
    bras=t%3
    gain=int(generateur.random()<vraies_probas[bras]);somme[bras]+=gain;comptes[bras]+=1
print('Valeurs estimées:',somme/comptes)
```

### Présentation

**Bandits manchots** (*Multi-Armed Bandits*) — famille : apprentissage par renforcement (sans état). **Objectif** : choisir répétitivement entre plusieurs actions (bras) pour maximiser la récompense cumulée en équilibrant exploration et exploitation.

**Intuition et étapes** : à chaque tour l'agent tire un bras, observe une récompense, met à jour l'estimation de la valeur de ce bras, puis décide du prochain tirage selon une stratégie. Dans l'exemple du notebook, les bras sont tirés en alternance pour estimer leurs probabilités de gain.

**Formule / critère essentiel** :

$$Q_t(a)=\frac{1}{N_t(a)}\sum_{s\le t}r_s\,\mathbb 1[a_s=a],\qquad R_T=T\mu^*-\sum_{t=1}^{T}r_t$$

**Symboles** : $a$ : bras ; $N_t(a)$ : nombre de tirages du bras $a$ ; $r_s$ : récompense au tour $s$ ; $Q_t(a)$ : valeur estimée ; $\mu^*$ : espérance du meilleur bras ; $T$ : horizon ; $R_T$ : regret cumulé.

**Paramètres appris** : les valeurs estimées $Q(a)$. **Hyperparamètres** : la stratégie de choix et ses paramètres (ex. $\varepsilon$, constante d'UCB), l'horizon $T$.

**Avantages** : (1) cadre simple pour étudier l'arbitrage exploration/exploitation ; (2) mise à jour légère, adaptée aux décisions en ligne (tests A/B, publicité).

**Limites** : (1) ne tient pas compte du contexte ni de l'état (sauf bandits contextuels) ; (2) suppose en général des récompenses stationnaires et indépendantes.

### Schéma explicatif

L'agent doit maximiser ses gains cumulés en choisissant entre trois bras ayant chacun des espérances de récompense différentes. Le schéma montre la mise à jour successive des estimations de récompense à travers l'historique d'interactions.

```python
fig, ax = plt.subplots(figsize=(8, 3.5))
ax.axis('off')
ax.set_xlim(-1, 9)
ax.set_ylim(-1, 3)

# Dessin des 3 bras du bandit
for i in range(3):
    ax.add_patch(plt.Rectangle((3*i, 0.5), 1.8, 1.5, fill=True, color='#cbd5e0', ec='black', lw=1.5))
    ax.text(3*i + 0.9, 1.25, f'Bras n°{i+1}\nE(R) = {vraies_probas[i]:.1f}', ha='center', va='center', weight='bold', fontsize=9)
    # Le levier
    ax.plot([3*i + 1.8, 3*i + 2.3], [1.25, 2.0], color='gray', lw=3)
    ax.add_patch(plt.Circle((3*i + 2.3, 2.0), 0.15, color='red'))

ax.text(4.5, -0.5, "L'agent doit arbitrer entre :\n1. Exploiter le bras actuellement estimé comme meilleur (Bras 3)\n2. Explorer les autres bras pour affiner ses estimations", ha='center', fontsize=9.5, style='italic', bbox=dict(facecolor='#feebc8', alpha=0.5))
ax.set_title("Bandits Manchots : Choix stratégiques de l'agent", fontsize=11)
plt.show()
```

2. **Allocation dynamique de serveurs de calcul (Logistique) :** Orienter dynamiquement les requêtes utilisateurs vers différentes configurations réseau ou serveurs physiques afin d'identifier la configuration la plus rapide tout en minimisant la latence globale.

**Cas où l'algorithme est peu approprié :**
Les bandits manchots classiques sont inadaptés lorsque les récompenses dépendent fortement d'un état d'environnement changeant d'une étape à l'autre (par exemple, des décisions d'achat dépendant du profil de l'utilisateur ou de l'historique récent de navigation). Dans ce cas, les modèles de décision markoviens ou les bandits contextuels sont indispensables.

## 26. Epsilon-greedy

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
generateur=np.random.default_rng(GRAINE)
vraies_probas=np.array([.2,.5,.7]);somme=np.zeros(3);comptes=np.zeros(3);epsilon=.1
for t in range(1000):
    bras=int(generateur.integers(3)) if generateur.random()<epsilon else int(np.argmax((somme+1)/(comptes+1)))
    gain=int(generateur.random()<vraies_probas[bras]);somme[bras]+=gain;comptes[bras]+=1
print('Nombre de choix:',comptes,'Valeurs estimées:',somme/np.maximum(comptes,1))
```

### Présentation

**Epsilon-greedy** (*Epsilon-greedy strategy*) — famille : apprentissage par renforcement (stratégie de bandit). **Objectif** : équilibrer exploration et exploitation par une règle de décision très simple.

**Intuition et étapes** : avec la probabilité $\varepsilon$ on choisit un bras au hasard (exploration), sinon on choisit le bras de plus forte valeur estimée (exploitation) ; on met à jour l'estimation après chaque récompense.

**Formule / critère essentiel** :

$$a_t=\begin{cases}\text{bras aléatoire}&\text{avec probabilité }\varepsilon\\ \arg\max_a Q_t(a)&\text{avec probabilité }1-\varepsilon\end{cases}$$

**Symboles** : $a_t$ : bras choisi au tour $t$ ; $\varepsilon\in[0,1]$ : probabilité d'explorer ; $Q_t(a)$ : valeur estimée du bras $a$.

**Paramètres appris** : les valeurs estimées $Q(a)$. **Hyperparamètres** : $\varepsilon$ (éventuellement décroissant dans le temps).

**Avantages** : (1) très simple à implémenter et à expliquer ; (2) garantit que tous les bras continuent d'être testés.

**Limites** : (1) exploration non dirigée : continue d'essayer des bras clairement mauvais ; (2) un $\varepsilon$ constant impose une perte de récompense permanente.

## 27. Q-learning

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False)
Q=np.zeros((env.observation_space.n,env.action_space.n));generateur=np.random.default_rng(GRAINE)
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode)
    for pas in range(100):
        action=int(generateur.integers(env.action_space.n)) if generateur.random()<max(.05,1-episode/2500) else int(np.argmax(Q[etat]))
        suivant,recompense,termine,tronque,_=env.step(action)
        cible=recompense+(0 if termine else .95*np.max(Q[suivant]))
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat=suivant
        if termine or tronque:break
print('Politique:',Q.argmax(axis=1));env.close()
```

### Présentation

**Q-learning** (*Q-learning*) — famille : apprentissage par renforcement (valeur, hors politique). **Objectif** : apprendre la valeur de chaque couple (état, action) pour en déduire une politique optimale, sans connaître le modèle de l'environnement.

**Intuition et étapes** : l'agent observe un état, choisit une action ($\varepsilon$-greedy), reçoit une récompense et un nouvel état, puis met à jour la table $Q$ avec la meilleure valeur possible dans l'état suivant ; la politique finale est l'action de plus forte valeur dans chaque état.

**Formule / critère essentiel** :

$$Q(s,a)\leftarrow Q(s,a)+\alpha\Big[r+\gamma\max_{a'}Q(s',a')-Q(s,a)\Big]$$

**Symboles** : $s,s'$ : état courant et suivant ; $a,a'$ : actions ; $r$ : récompense ; $\alpha$ : taux d'apprentissage ; $\gamma$ : facteur d'actualisation ; $Q$ : valeur d'une action dans un état.

**Paramètres appris** : la table $Q(s,a)$. **Hyperparamètres** : $\alpha$, $\gamma$, $\varepsilon$ (et sa décroissance), nombre d'épisodes.

**Avantages** : (1) converge vers la politique optimale (conditions classiques) même en explorant ; (2) simple et sans modèle de l'environnement.

**Limites** : (1) version tabulaire inutilisable avec de nombreux états ou des états continus ; (2) sensible aux hyperparamètres, peut surestimer les valeurs.

## 28. SARSA

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False);Q=np.zeros((16,4));generateur=np.random.default_rng(GRAINE)
def choisir(etat,epsilon):
    return int(generateur.integers(4)) if generateur.random()<epsilon else int(np.argmax(Q[etat]))
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode);epsilon=max(.05,1-episode/2500);action=choisir(etat,epsilon)
    for pas in range(100):
        suivant,recompense,termine,tronque,_=env.step(action)
        action_suiv=choisir(suivant,epsilon)
        cible=recompense+(0 if termine else .95*Q[suivant,action_suiv])
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat,action=suivant,action_suiv
        if termine or tronque:break
print('Politique SARSA:',Q.argmax(axis=1));env.close()
```

### Présentation

**SARSA** (*State-Action-Reward-State-Action*) — famille : apprentissage par renforcement (valeur, sur la politique). **Objectif** : apprendre la valeur des actions de la politique effectivement suivie, exploration comprise.

**Intuition et étapes** : comme pour le Q-learning, mais la mise à jour utilise l'action $a'$ réellement choisie dans l'état suivant (selon la même politique $\varepsilon$-greedy) et non l'action optimale.

**Formule / critère essentiel** :

$$Q(s,a)\leftarrow Q(s,a)+\alpha\Big[r+\gamma\,Q(s',a')-Q(s,a)\Big]$$

**Symboles** : $s,s'$ : état courant et suivant ; $a,a'$ : action choisie et action suivante réellement choisie ; $r$ : récompense ; $\alpha$ : taux d'apprentissage ; $\gamma$ : facteur d'actualisation.

**Paramètres appris** : la table $Q(s,a)$. **Hyperparamètres** : $\alpha$, $\gamma$, $\varepsilon$ (et sa décroissance), nombre d'épisodes.

**Avantages** : (1) tient compte du risque lié à l'exploration, politique plus prudente ; (2) stable, adapté aux environnements où les erreurs coûtent cher.

**Limites** : (1) la politique apprise dépend de $\varepsilon$ et peut être moins performante que l'optimale ; (2) également tabulaire, inadapté aux grands espaces d'états.

## 29. DQN — cible conceptuelle

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
# Calcul de cible DQN, sans entraîner un réseau neuronal complet
recompenses=np.array([0.,1.,-1.]);termine=np.array([False,True,False]);meilleur_Q_suivant=np.array([2.,3.,1.]);gamma=.95
cibles=recompenses+gamma*(~termine)*meilleur_Q_suivant
print('Cibles de Bellman:',cibles)
# Un DQN complet ajoute réseau en ligne, réseau cible et mémoire de rejeu.
```

### Présentation

**Réseau Q profond** (*Deep Q-Network, DQN*) — famille : apprentissage par renforcement (valeur, approximation par réseau de neurones). **Objectif** : étendre le Q-learning aux grands espaces d'états en approchant la fonction $Q$ par un réseau de neurones.

**Intuition et étapes** : l'agent stocke ses transitions dans une mémoire de rejeu, tire des mini-lots aléatoires, calcule une cible de Bellman avec un réseau cible figé, et entraîne le réseau principal pour s'en rapprocher ; le réseau cible est recopié périodiquement. L'exemple du notebook ne calcule que cette cible.

**Formule / critère essentiel** :

$$y=r+\gamma(1-d)\max_{a'}Q(s',a';\theta^-),\qquad L(\theta)=\mathbb E\big[(y-Q(s,a;\theta))^2\big]$$

**Symboles** : $\theta$ : poids du réseau principal ; $\theta^-$ : poids du réseau cible ; $d\in\{0,1\}$ : 1 si l'épisode est terminé ; $r,\gamma$ : récompense et facteur d'actualisation ; $y$ : cible ; $L$ : perte quadratique.

**Paramètres appris** : les poids $\theta$ du réseau (et la copie $\theta^-$). **Hyperparamètres** : taux d'apprentissage, $\gamma$, $\varepsilon$, taille de la mémoire et du mini-lot, fréquence de mise à jour du réseau cible.

**Avantages** : (1) gère des états de grande dimension (images, capteurs) ; (2) mémoire de rejeu et réseau cible stabilisent l'apprentissage.

**Limites** : (1) gourmand en données et en calcul, entraînement parfois instable ; (2) très sensible aux hyperparamètres, difficile à reproduire.

### Schéma explicatif

**Comment lire ce schéma :** Le diagramme montre l'architecture fonctionnelle d'un **DQN (Deep Q-Network)**. Pour stabiliser l'apprentissage d'un réseau neuronal profond confronté à des données corrélées, le DQN utilise une **Mémoire de rejeu** (qui casse la corrélation temporelle en mélangeant les transitions) et un **Réseau Cible** (dont les poids $\theta^-$ sont figés temporairement pour éviter les oscillations de la cible de Bellman).

```python
# Régression linéaire par descente de gradient : données univariées
x=np.linspace(-1,1,100); y=2+3*x
poids=0.;biais=0.;pas=.1
for iteration in range(500):
    prediction=poids*x+biais
    erreur=prediction-y
    poids-=pas*2*np.mean(erreur*x)
    biais-=pas*2*np.mean(erreur)
print('Paramètres appris manuellement:',poids,biais)
from sklearn.linear_model import LinearRegression
print('Référence sklearn:',LinearRegression().fit(x.reshape(-1,1),y).coef_,LinearRegression().fit(x.reshape(-1,1),y).intercept_)
```

```python
# K-means manuel : assignation et recalcul jusqu'à stabilité
valeurs=X_clu.copy(); centroids=valeurs[np.random.default_rng(GRAINE).choice(len(valeurs),size=3,replace=False)].copy()
for iteration in range(100):
    distances=((valeurs[:,None,:]-centroids[None,:,:])**2).sum(axis=2)
    groupes=distances.argmin(axis=1)
    nouveaux=np.array([valeurs[groupes==k].mean(axis=0) if (groupes==k).any() else centroids[k] for k in range(3)])
    if np.allclose(nouveaux,centroids):break
    centroids=nouveaux
print('Centroïdes manuels:',centroids)
from sklearn.cluster import KMeans
print('Centroïdes sklearn (ordre arbitraire):',KMeans(n_clusters=3,n_init=10,random_state=GRAINE).fit(valeurs).cluster_centers_)
```
